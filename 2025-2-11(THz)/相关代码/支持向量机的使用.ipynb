```python
import numpy as np
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix

'''
数据格式如下：
第1列是THz值（不使用），第2-10列为化合物一的吸收强度，第11列为化合物一的标签
第12-20列为化合物二的吸收强度，第21列为化合物二的标签
以此类推，直到第201列
'''

# 提取训练集和测试集
# 初始化特征和标签列表
X = []
y = []

# 遍历每个化合物
for i in range(20):
    # 特征列索引
    feature_start = 1 + i * 10  # 每个化合物的特征起始位置
    feature_end = feature_start + 9  # 10列吸收强度
    label_index = feature_end  # 测试集标签

    # 提取特征和标签
    X.append(data.iloc[:, feature_start:feature_end].values)  # 吸收强度
    y.append(data.iloc[:, label_index].values)                # 标签
    
    

# 将特征和标签合并为数组
X = np.vstack(X)
y = np.concatenate(y)

# 如果 y 是字符串类型的标签，使用 LabelEncoder 进行编码
if y.dtype.kind in 'biuf':  # 检查是否为数值类型
    y = np.array(y, dtype=int)  # 转换为整型（如果需要）

# 创建并训练SVM模型
model = SVC(kernel='linear')  # 你可以选择不同的kernel，例如'rbf'
model.fit(X, y)

# 进行预测
y_pred = model.predict(X)

# 输出评估结果
print(confusion_matrix(y, y_pred))
print(classification_report(y, y_pred))
```
