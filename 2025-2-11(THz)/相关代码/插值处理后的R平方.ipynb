```python
from sklearn.metrics import r2_score
import pandas as pd

# 这个是整合后的数据（1-3 THz）
df = pd.read_excel("插值整合后的数据.xlsx", sheet_name='Sheet1')

# 这里是创建存储R方值的字典
rfangzidian = {}

# 循环计算 R² 值
for i in range(20):
    observed_col = df.iloc[:, i * 10 + 1]  # 第 1, 11, 21, ... 列
    fitted_col = df.iloc[:, i * 10 + 10]  # 第 10, 20, 30, ... 列

    # 计算 R² 值
    rfang = r2_score(observed_col, fitted_col)

    # 存储结果
    rfangzidian[f'huahewu{i + 1}'] = rfang

# 打印结果
for compound, r2 in rfangzidian.items():
    print(f"{compound} 的 R² 值为: {r2}")
```
