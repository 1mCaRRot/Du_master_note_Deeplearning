```julia
using Plots

# 参数设置
λ = 0.3  # 接触率
μ = 0.1  # 治愈率
N = 1000  # 总人口
I0 = 0.01 * N  # 初始感染者
R0 = 0         # 初始恢复者
S0 = 0.99 * N  # 初始健康者

# 时间设置
t_total = 160  # 总时间
dt = 1         # 时间步长
time = 0:dt:t_total

# 初始化S, I, R数组
S = zeros(length(time))
I = zeros(length(time))
R = zeros(length(time))

# 初始值
S[1] = S0
I[1] = I0
R[1] = R0

# SIR模型的迭代
for t in 1:(length(time)-1)
    dS = -λ * S[t] * I[t] / N * dt
    dI = (λ * S[t] * I[t] / N - μ * I[t]) * dt
    dR = μ * I[t] * dt

    S[t + 1] = S[t] + dS
    I[t + 1] = I[t] + dI
    R[t + 1] = R[t] + dR
end

# 计算比例
S_ratio = S ./ N
I_ratio = I ./ N
R_ratio = R ./ N

# 绘制结果
plot(time, S_ratio, label="健康者比例 (S/N)", color=:blue)
plot!(time, I_ratio, label="感染者比例 (I/N)", color=:red)
plot!(time, R_ratio, label="恢复者比例 (R/N)", color=:green)
xlabel!("时间")
ylabel!("比例")
title!("SIR模型的健康者、感染者与恢复者比例")
legend!
```
