```julia
using DifferentialEquations
using Plots
```

```julia
# 模型参数
N = 2521  # 总人口
I0 = 339  # 初始感染者
R0 = 5    # 初始康复者
S0 = N - I0 - R0  # 初始易感者

# 传染率和恢复率，假设值，这里根据PPT的内容将接触率设为0.6，治愈率设为0.3（变量名更改了一下）
beta = 0.6  # 接触率
gamma = 0.3  # 治愈率
```

```julia
function sir!(du, u, p, t)
    S, I, R = u
    β, γ = p
    du[1] = -β * S * I / N  # 易感者变化
    du[2] = β * S * I / N - γ * I  # 感染者变化
    du[3] = γ * I  # 康复者变化
end
```

```julia
u0 = [S0, I0, R0]  # 初始条件
tspan = (0.0, 100.0)  # 模拟时间范围
p = [beta, gamma]  # 参数
```

```julia
# 定义问题
prob = ODEProblem(sir!, u0, tspan, p)

# 求解
sol = solve(prob)
```

```julia
# 设置字体
gr(fontfamily="Arial Unicode MS")  # 或者使用其他支持中文的字体

#绘图
plot(sol, label=["易感者" "感染者" "康复者"], xlabel="时间", ylabel="人数", legend=:topright)
```

```julia
# 1. 导入所需要的库
using DifferentialEquations
using Plots

# 2. 定义模型参数
N = 2521  # 总人口
I0 = 339  # 初始感染者
R0 = 5    # 初始康复者
S0 = N - I0 - R0  # 初始易感者

# 传染率和恢复率，假设值，这里根据PPT的内容将接触率设为0.6，治愈率设为0.3（变量名更改了一下）
beta = 0.6  # 接触率
gamma = 0.3  # 治愈率

# 3. 定义微分方程
function sir!(du, u, p, t)
    S, I, R = u
    β, γ = p
    du[1] = -β * S * I / N  # 易感者变化
    du[2] = β * S * I / N - γ * I  # 感染者变化
    du[3] = γ * I  # 康复者变化
end

# 4. 设置初始条件和时间范围
u0 = [S0, I0, R0]  # 初始条件
tspan = (0.0, 100.0)  # 模拟时间范围
p = [beta, gamma]  # 参数

# 5. 求解微分方程
# 定义问题
prob = ODEProblem(sir!, u0, tspan, p)
# 求解
sol = solve(prob)

# 6. 图像可视化（使用中文，所以先设置字体）
# 设置字体
gr(fontfamily="Arial Unicode MS")  # 或者使用其他支持中文的字体
#绘图
plot(sol, label=["易感者" "感染者" "康复者"], xlabel="时间", ylabel="人数", legend=:topright)
# 保存图形
savefig("SIR模型绘图.png")  # 您可以选择其他格式，如 .pdf, .svg 等
```
